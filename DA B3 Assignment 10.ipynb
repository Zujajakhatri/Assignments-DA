```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
```

```python
q = '''select * from Salaries'''
sql_query(q)
```

1. Top 10 highest paid employees (by TotalPayBenefits)

```python
q= '''SELECT EmployeeName, JobTitle, TotalPayBenefits
FROM Salaries
ORDER BY TotalPayBenefits DESC
LIMIT 10'''

sql_query(q)

```

2. Find which year has missing data of Employees with 'Not Provided'/ 'Not provided' .

```python
q = '''select * from Salaries
    where year == "Not provided" 
    '''

sql_query(q)
```

3. Number of unique agencies

```python
q= '''SELECT COUNT(DISTINCT Agency) AS UniqueAgencyCount
FROM Salaries'''

sql_query(q)
```

4. Show Average BasePay, OvertimePay, and TotalPay for year 2011

```python
q= '''select 
    avg(basepay) as average_basepay,
    avg(overtimepay) as average_overtimepay,
    avg(totalpay) as average_totalpay,
    year
from salaries
where year = 2011
'''

sql_query(q)
```

5. Plot the top 10 job titles with the highest TotalPay in 2014.

```python
q= '''select jobtitle, avg(totalpay) as AverageTotalPay, year
    from salaries
    where year = 2014
    group by jobtitle
    order by AverageTotalPay desc
    limit 10;
'''
sql_query(q)
```

6. Which job titles have overtime pay exceeding $150,000 exluding 'Not provided'

```python
q= '''select jobtitle, employeename, overtimepay
    from salaries
    where overtimepay > 150000
    and jobtitle != 'not provided'
'''
sql_query(q)

```

7. Find all employees whose job titles contain the word “CHIEF”

```python
q= '''select employeename, jobtitle
    from salaries
    where jobtitle like '%chief%' collate nocase;
'''

sql_query(q)
```

8. Count how many employees have job titles containing “ENGINEER” and “Fire” in year 2012 and 2013

```python
q= '''select count(*) as employee_count
    from salaries
    where (jobtitle like '%engineer%' and jobtitle like '%fire%')
    and year in (2012, 2013)
    collate nocase'''

sql_query(q)
```

9. Distinct job titles starting with "CAPTAIN"

```python
q= '''select distinct jobtitle
    from salaries
    where jobtitle like 'captain%' collate nocase
'''
sql_query(q)
```

10. Employees having basepay > 50k and totalpay < 150000 exlucding 'Not Provided'

```python
q= '''select employeename, jobtitle, basepay, totalpay
    from salaries
    where basepay != 'not provided'
    and cast(basepay as real) > 50000
    and totalpay < 150000
    '''

sql_query(q)
```
